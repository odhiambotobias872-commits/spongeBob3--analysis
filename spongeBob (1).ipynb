```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


data = {
    "Name": ["SpongeBob", "patrick", "Squidward", "Mr Krabs", "Sandy", "Plankton", "Gary", "Mr Puff"],
    "Department": ["Kitchen", "Kitchen", "Cashier", "Management", "Science", "Science", "Kitchen", "Boating"],
    "Score": [ 95, np.nan, 70, 88, np.nan, 45, 33, 72],
    "Hours": [40, 20, 35, 50, 45, 10, np.nan, 38]

}
df = pd.DataFrame(data)
print(df)

# add a column age
df["Age"] = [35, 36, 37, 38, 39, 31, 30, 33]
print(df)

# nan in each column
df.isnull().sum()

# fill nan scores with meaan
print(df["Score"].mean())
df["Score"] = (df["Score"].fillna(df["Score"].mean()))
print(df["Score"].mean())


 # fill nan hours with 0
df["Hours"] = df["Hours"].fillna(0)

print(df)
print(df.isnull().sum())

# finding std deviation and variance

print("std deviation:", df["Score"].std())
print("variance:", df["Score"].var())

# finding max and min
print("minimum:", df["Score"].min())

print("maximum:", df["Score"].max())

print("---Sorted by Score---")
print(df.sort_values(by="Score", ascending = False))

print(df.groupby("Department")["Score"].mean()) # which department performs best
print(df["Hours"].corr(df["Score"])) # more hours = higher score

# creating a data frame
df.plot(kind="scatter", x="Hours", y="Score", title = "High hours; High score")
plt.savefig("1_scatter_hours_us_score.png")
plt.show()


df.plot(kind="bar", x = "Name", title = "High Score", legend=False)
plt.xticks(rotation=45)
plt.ylabel(["Score", "Hours", "Age"])
plt.tight_layout()
plt.savefig("2_bar_name_score.png")
plt.show()

df.groupby("Department")["Score"].mean().plot(kind="bar", title="Average Score by Department")
plt.ylabel("Average Score")
plt.xlabel("Department")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("3_bar_department_average.png")
plt.show()









```

```python
import random

secret = random.randint(20, 25)
attempts = 0

while attempts < 3:
  print("guess a number between 20 and 25")
  guess = int(input("enter a number between 20 and 25: "))

  if guess > secret:
    print("Too High! Please try a lower number.")

  elif guess < secret:
    print("Oops! Too low!")

  else:
    print("You got it! You are a genius!")
    break

  attempts += 1

  if attempts == 3 and guess != secret:
    print(f"Game over! The secret number was {secret}.")



```
